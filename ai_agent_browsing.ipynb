```python
import os
import google.generativeai as genai
from ddgs import DDGS

# 1. API Key Setup (Apni Gemini Key yahan lagaein agar pehle set nahi ki)
os.environ["GEMINI_API_KEY"] = "YOUR_GEMINI_API_KEY_HERE"
genai.configure(api_key=os.environ["GEMINI_API_KEY"])

# Model setup
model = genai.GenerativeModel('gemini-1.5-flash')

# 2. Web Search Function
def search_web(query: str):
    with DDGS() as ddgs:
        results = list(ddgs.text(query, max_results=3))
        formatted_results = "\n\n".join([f"Title: {r['title']}\nSnippet: {r['body']}\nURL: {r['href']}" for r in results])
        return formatted_results

# 3. Autonomous Browsing Agent Class
class AutonomousBrowsingAgent:
    def __init__(self, model, search_func):
        self.model = model
        self.search_func = search_func

    def run(self, user_query: str):
        print(f"🤖 [Agent]: Processing User Query -> '{user_query}'")
        
        # Step A: Reasoning & Query Planning
        planner_prompt = f"""
        You are an AI Search Planner. 
        User Question: "{user_query}"
        
        Formulate a short, effective web search query to find accurate and latest information.
        Return ONLY the search query text, no quotes or additional prose.
        """
        
        search_query = self.model.generate_content(planner_prompt).text.strip()
        print(f"🔍 [Agent Thinking]: Optimized Search Query -> '{search_query}'")
        
        # Step B: Tool Execution
        try:
            print("🌐 [Agent Action]: Browsing the Web...")
            search_results = self.search_func(search_query)
            print("✅ [Agent Action]: Real-time Data Fetched!")
        except Exception as e:
            search_results = f"Search Failed due to error: {str(e)}"
            
        # Step C: Information Synthesis & Final Response
        final_prompt = f"""
        You are an intelligent Autonomous AI Assistant with web access. 
        Answer the user's query comprehensively using the live search data below.
        
        User Query: {user_query}
        
        Live Web Search Context:
        {search_results}
        
        Formatting Guidelines:
        - Provide a clear, well-structured, and concise answer.
        - Use bullet points for key takeaways.
        - Mention sources/URLs inline if helpful.
        """
        
        final_response = self.model.generate_content(final_prompt).text
        return final_response

# Agent Initialize Karein
agent = AutonomousBrowsingAgent(model=model, search_func=search_web)
print("🚀 Agent is Ready to use!")
```
